```python
#1.3

import matplotlib.pyplot as plt

years = []
rates = []
colors = []

threshold = 45  # 45% threshold

# Calculate graduation rates
for record in vital_stats:
    year = record["year"]
    rate = (record["graduated"] / record["enrolled"]) * 100

    years.append(year)
    rates.append(rate)

    # Assign color based on threshold
    if rate >= threshold:
        colors.append("green")   # meets/exceeds
    else:
        colors.append("yellow")     # below

# Create bar chart
plt.figure()
plt.bar(years, rates, color=colors)

# Add threshold line
plt.axhline(y=threshold, linestyle='--')

# Labels and title
plt.xlabel("Year")
plt.ylabel("Graduation Rate (%)")
plt.title("Graduation Rate Analysis (Threshold = 45%)")

# Optional: show values on top of bars
for i in range(len(years)):
    plt.text(years[i], rates[i], f"{rates[i]:.1f}%", ha='center')

# Show plot
plt.show()
```

```python
#1.4

import matplotlib.pyplot as plt


# Reusable function
def contribution_percentage(data, selected_year):
    total_enrolled = sum(d["enrolled"] for d in data)
    total_graduated = sum(d["graduated"] for d in data)

    for d in data:
        if d["year"] == selected_year:
            enrol_pct = (d["enrolled"] / total_enrolled) * 100
            grad_pct = (d["graduated"] / total_graduated) * 100
            return enrol_pct, grad_pct

    print("Year not found.")
    return None, None


# Example usage (demonstration)
year = 2020
enrol_pct, grad_pct = contribution_percentage(vital_stats, year)

# Visualization
labels = ["Enrolment %", "Graduation %"]
values = [enrol_pct, grad_pct]

plt.figure()
plt.bar(labels, values)

plt.title(f"Percentage Contribution for Year {year}")
plt.ylabel("Percentage (%)")

# Add value labels
for i in range(len(values)):
    plt.text(i, values[i], f"{values[i]:.2f}%", ha='center')

plt.show()

# Print results
print(f"Year: {year}")
print(f"Enrolment Contribution: {enrol_pct:.2f}%")
print(f"Graduation Contribution: {grad_pct:.2f}%")
```

```python
#2.3

# 2D dataset
def transfer_contribution(data):
    total_transfer = 0
    total_students = 0

    for row in data:
        transfer = row[3]           # Transfer is index 3
        row_total = sum(row)        # Total students in that year

        total_transfer += transfer
        total_students += row_total

    # Calculate percentage contribution
    contribution = (total_transfer / total_students) * 100

    return total_transfer, total_students, contribution


# Run calculation
transfer_total, overall_total, percentage = transfer_contribution(data)

# Display results
print(f"Total Transfer Students: {transfer_total:,}")
print(f"Total Students (All Categories): {overall_total:,}")
print(f"Transfer Contribution: {percentage:.2f}%")
```

```python
import math

class AcademicStaffStats:
    # Shared sector information (accessible without any instance)
    SECTOR = "South African Public Higher Education Sector"

    # Reasonable target stability level used for "what-if" change calculations
    TARGET_STABILITY_PERCENT = 70.0

    def __init__(self, year: int, permanent: int, temporary: int):
        #Year
        if not isinstance(year, int) or year < 2000 or year > 2100:
            raise ValueError("Reporting year must be an integer between 2000 and 2100.")
        #Permanent Staff count
        if not isinstance(permanent, int) or permanent < 0:
            raise ValueError("Permanent staff count must be a non-negative integer.")
        #Temp Staff count
        if not isinstance(temporary, int) or temporary < 0:
            raise ValueError("Temporary staff count must be a non-negative integer.")
            
        # Prevent zero-total objects
        self.total = permanent + temporary
        if self.total == 0:
            raise ValueError("Total academic staff cannot be zero. At least one staff member is required.")
            
        #storing Perma and temp staff counts
        self.year = year
        self.permanent = permanent
        self.temporary = temporary

    @classmethod
    def get_sector(cls) -> str:
        """Retrieve the shared sector name without needing an object instance."""
        return cls.SECTOR

    # ────────────────────────────────────────────────
    # Core calculations (all protected against invalid states)
    # ────────────────────────────────────────────────
    def total_staff(self) -> int:
        """Total academic staff for the year (always safe)."""
        return self.total

    def temporary_percentage(self) -> float:
        """Percentage share of temporary staff (rounded to 1 decimal)."""
        return round((self.temporary / self.total) * 100, 1)

    def stability_indicator(self) -> str:
        """Staff stability indicator: proportion of permanent staff (meaningful reporting form)."""
        perm_pct = round((self.permanent / self.total) * 100, 1)
        return f"{perm_pct}% permanent staff"

    def temporary_to_permanent_ratio(self) -> str:
        """
        Temporary-to-permanent ratio (or equivalent comparative measure).
        Handles exceptional cases (e.g. zero permanent staff) without silent errors.
        """
        if self.permanent == 0:
            if self.temporary == 0:
                return "0:1 (no staff)"
            return "Infinite (all temporary – no permanent staff)"

        ratio = round(self.temporary / self.permanent, 2)
        return f"{ratio}:1 (temporary staff per permanent staff)"

    def required_change_to_target(self) -> str:
        """
        "What would need to change?" calculation.
        Returns how many positions must shift between categories to reach
        the target stability level (70% permanent). Handles all edge cases.
        """
        target = self.TARGET_STABILITY_PERCENT
        target_perm = round((target / 100) * self.total)
        shift_needed = target_perm - self.permanent

        if shift_needed == 0:
            return f"Already at target stability ({target}% permanent). No change needed."

        if shift_needed > 0:
            # Need to convert temporary → permanent
            if shift_needed > self.temporary:
                return f"Impossible with current data – would require {shift_needed:,} shifts but only {self.temporary:,} temporary staff exist."
            direction = "from temporary to permanent"
            positions = shift_needed
        else:
            # Would need to convert permanent → temporary (rarely desired)
            direction = "from permanent to temporary"
            positions = abs(shift_needed)

        return (f"To reach {target}% permanent staff, "
                f"{positions:,} positions would need to shift {direction}.")

    def generate_report(self) -> None:
        """Print a clear, professional workforce statistics report."""
        print("═" * 70)
        print(f"ACADEMIC STAFF WORKFORCE REPORT – {self.get_sector().upper()}")
        print("═" * 70)
        print(f"Reporting Year                  : {self.year}")
        print(f"Total Academic Staff            : {self.total_staff():,}")
        print(f"Temporary Staff Percentage      : {self.temporary_percentage()}%")
        print(f"Staff Stability Indicator       : {self.stability_indicator()}")
        print(f"Temporary-to-Permanent Ratio    : {self.temporary_to_permanent_ratio()}")
        print(f"Required Change to Target ({self.TARGET_STABILITY_PERCENT}% permanent):")
        print(f"   → {self.required_change_to_target()}")
        print("═" * 70)


# ────────────────────────────────────────────────
# Demonstration using official CHE 2021 statistics
# ────────────────────────────────────────────────
if __name__ == "__main__":
    print("Creating AcademicStaffStats object for 2021 (CHE data)...\n")

    try:
        stats_2021 = AcademicStaffStats(year=2021, permanent=54000, temporary=30500)
        stats_2021.generate_report()

        # Also demonstrate class-level sector access (no object needed)
        print(f"\nSector information retrieved directly from class: {AcademicStaffStats.get_sector()}")

    except ValueError as e:
        print(f"Validation error: {e}")
```
